```python
# vscode demo - Design Patterns

# 01-reflection-pattern; agent, main, reflection
# Python self = Java this

# agent.py - request -> response
class Agent:
  def __init__(self):
    self.attempt = 0 # globally declared


  def act(self, task: str) -> str: # str = method return type
    self.attempt += 1

    # perform activity
    if self.attempt == 1:
      return "AI is cool. "
    elif self.attempt == 2:
      return "AI is awesome"
    else:
      return(
          "AI is the best and very powerful"
          "AI is the future because")
```

```python
# Reflector
class Reflector:
  def reflect(self, task, output: str) -> str:
    if len(output) < 15:
      return "too short, needs more detail"
    if "because" not in output:
      return "explanation lacks reasoning"
    return "good answer"
```

```python
agent = Agent()
reflector = Reflector()

Task = "Explain why AI is important."

response = ""
reflection_feedback = ""

max_attempts = 3

# loop until get good answer
for step in range(max_attempts):
  print(f"\n Iteration {step + 1} ----")

# decide tool -- > calculator

  output = agent.act(Task)
  print(f"Agent output: {output}")

  reflection_feedback = reflector.reflect(Task, output)
  print(f"Reflector feedback: {reflection_feedback}")

  if reflection_feedback == "good answer":
    print("Task is completed successfully")
    break
```

```python
# 1. Install/Upgrade the consolidated package
# !pip install -qU langchain-google-genai faiss-cpu

import os
import sys
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Use the consolidated package imports
from langchain_google_genai import GoogleGenerativeAIEmbeddings, ChatGoogleGenerativeAI

from langchain_community.vectorstores import FAISS
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 2. Authenticate the Notebook Session
if "google.colab" in sys.modules:
    from google.colab import auth
    print("Authenticating user session...")
    auth.authenticate_user()

# Define Google Cloud Configuration
PROJECT_ID = "fc26-mi-defense-prod"  # Your Google Cloud Project ID
LOCATION = "us-central1"            # Region hosting your Vertex AI models

# Create sample data.txt if it doesn't exist
if not os.path.exists("data.txt"):
    with open("data.txt", "w", encoding="utf-8") as f:
        f.write("Agentic AI design patterns are key systems architectures for building modern LLM systems. "
                "There are six core patterns: Planning, Reflection, Tool Use, ReAct, ReWOO, and Multi-Agent Collaboration.")

# Load and split documents
loader = TextLoader(file_path="data.txt", encoding="utf-8")
documents = loader.load()

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
docs = text_splitter.split_documents(documents)

# 3. Configure Embeddings for Vertex AI (using GoogleGenerativeAIEmbeddings)
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/text-embedding-004",  # Recommended embedding model
    project=PROJECT_ID,
    location=LOCATION,
    vertexai=True                       # Instructs the SDK to use Vertex AI, not AI Studio
)

# 4. Generate Vector Store
vectorstore = FAISS.from_documents(documents=docs, embedding=embeddings)
retriever = vectorstore.as_retriever()

# 5. Configure Chat Model for Vertex AI (using ChatGoogleGenerativeAI)
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",      # Explicit version resolves 404 endpoint mapping issues
    project=PROJECT_ID,
    location=LOCATION,
    vertexai=True                       # Instructs the SDK to use Vertex AI, not AI Studio
)

prompt = ChatPromptTemplate.from_template("Answer the question based only on the context:\n\n{context}\n\nQuestion: {question}")

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {
        "context": (lambda x: x["question"]) | retriever | format_docs,
        "question": lambda x: x["question"]
    }
    | prompt
    | llm
    | StrOutputParser()
)

query = "What is the doc all about?"

try:
    result = rag_chain.invoke({"question": query})
    print("\n--- SUCCESSFUL VERTEX AI PIPELINE RUN ---")
    print(result)

    print("\n--- Query 2 ---")
    print(rag_chain.invoke({"question": "What are the core patterns in Agentic AI design?"}))
except Exception as e:
    print(f"\nAn error occurred during execution: {e}")
```
